```python
import pandas as pd
df=pd.read_csv('orders_practice.csv')
df.columns
```

```python
Missing_age=df['Age'].isna()
display(df[Missing_age])
display(len(df[Missing_age]))
```

```python
Missing_Rating_City=(df['Rating'].isna())|(df['City'].isna())
df[Missing_Rating_City]
```

```python
wrong_age = (df['Age'] < 18) | (df['Age'] > 100)
df[wrong_age]
```

```python
Not_Rating_age=(df['Rating'].notna())&(df['Age'].notna())
df[Not_Rating_age]
```

```python
good_order=(df['Total Amount']>10000) & (df['Rating']>=4) & (df['City'].notna())
display(df[good_order])
print(good_order.sum())
```
